```python
import pandas as pd
df=pd.read_csv(r"C:\Users\Admin\Downloads\Titanic-Dataset.csv")
df
```

```python
import matplotlib.pyplot as plt
counts=df["Pclass"].value_counts().sort_index()
plt.bar(counts.index , counts.values)
plt.xlabel("passenger class")
plt.ylabel("number of passengers")
plt.title("number of passengers in each class")
plt.xticks([1,2,3])
plt.show
```

```python
counts=df.groupby(["Survived","Sex"]).size().unstack()
print(counts)
counts.plot(kind="bar")
plt.xlabel("survival status")
plt.ylabel("number of passengers")
plt.title("titanic survival by sex")
plt.xticks([0,1],["not survived","survived"] , rotation=0)
plt.show
```

```python
data=df.dropna(subset=["Age","Fare"])
plt.scatter(data["Age"],data["Fare"])
plt.xlabel("age")
plt.ylabel("fare")
plt.title("age vs fare in titanic data set")
plt.show()
```

```python
data=df.dropna(subset=["Pclass","Fare"])
plt.scatter(data["Pclass"],data["Fare"])
plt.xlabel("passenger class")
plt.ylabel("fare")
plt.title("passenger class vs fare in titanic data set")
plt.xticks([1,2,3])
plt.grid(True)
plt.show()
```

```python
data=df.dropna(subset=["Pclass","Age"])
plt.scatter(data["Pclass"],data["Age"])
plt.xlabel("passenger class")
plt.ylabel("age")
plt.title("age vs passenger class in titanic data set")
plt.xticks([1,2,3])
plt.grid(True)
plt.show()
```

```python
avg_fare=df.groupby(["Pclass"],["Fare"].mean())
plt.plot(avg_fare.index, avg_fare.values , marker="o")
plt.xlabel("passemger class")
plt.ylabel("fare")
plt.title("avg fare vs passenger class")
plt.xticks
```
